```python
import pandas as pd
from sklearn.model_selection import train_test_split
from pathlib import Path
```

```python
RANDOM_SEED = 29
RAW_TRAIN_PATH = Path("../data/raw/train_raw.csv")
PROC_TRAIN_PATH = Path("../data/processed/train.csv")
PROC_VAL_PATH = Path("../data/processed/val.csv")
```

```python
df = pd.read_csv(RAW_TRAIN_PATH)
```

```python
train_df, val_df = train_test_split(
    df, 
    test_size=0.2, 
    random_state=RANDOM_SEED, 
    stratify=df['Exited'] 
)
```

```python
train_df.to_csv(PROC_TRAIN_PATH, index=False)
val_df.to_csv(PROC_VAL_PATH, index=False)
```

```python
print(f"Данные успешно разделены!")
print(f"Train: {train_df.shape}, Доля оттока: {train_df['Exited'].mean():.2%}")
print(f"Val:   {val_df.shape}, Доля оттока: {val_df['Exited'].mean():.2%}")
```
